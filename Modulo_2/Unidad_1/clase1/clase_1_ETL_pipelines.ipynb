## Extract: Multiples Archivos (con glob)

/data/ventas/
- ventas_2026_01.csv
- ventas_2026_02.csv
- ventas_2026_03.csv

```python
def extract_multiples_archivos(patron_glob):
    
    archivos = glob.glob(patron_glob)
    
    if not archivos:
        print(f' No se encontraron archivos con el patrón: {patron_glob}')
        return pd.DataFrame()
    
    print(f" Archivos encontrados: {len(archivos)}")
    dfs = []
    
    for ruta in sorted(archivos):
        df_tmp = pd.read_csv(ruta)
        
        df_tmp['archivo_origen'] = Path(ruta).name
        
        dfs.append(df_tmp)
        print(f" {Path(ruta).name}: {len(df_tmp)} filas")
        
    df_combinado = pd.concat(dfs, ignore_index=True)
    return df_combinado

#Simulación: 
os.makedirs('data_lab', exist_ok=True)

datos_mes = {
    "data_lab/empleados_2024_01.csv": """id,nombre,salario
1,Alice,3000
2,Bob,3500
3,Carol,4000
4,David,2800""",

    "data_lab/empleados_2024_02.csv": """id,nombre,salario
3,Carol,4000
4,David,2800
5,Eve,5000
6,Frank,3200""",

    "data_lab/empleados_2024_03.csv": """id,nombre,salario
5,Eve,5000
6,Frank,3200"""
}

for ruta, contenido in datos_mes.items():
    with open(ruta, 'w') as f:
        f.write(contenido)
        
print("=== Leyendo multiples archivos con glob ====")
df_multi = extract_multiples_archivos("data_lab/empleados_2024_*.csv")

print(f"Dataframe combinado: {df_multi.shape}")
df_multi
```

## Extract: Consumo de APIs externas



¿Qué es una API REST?

Es un servidor que expone datos a través de URLs. 



Patrones Críticos:

Paginación:

Rate limiting:

Manejo de errores: 

Reintentos: 

LA funcion clave: request.get()

```python
def extract_api(base_url, params_base=None, max_paginas=3, pausa_entre_pagina=0.5):
    
    params_base = params_base or {}
    todos_registros = []
    
    for pagina in range(1, max_paginas + 1):
        params = {
            **params_base,
            '_start': (pagina - 1) * 10,
            '_limit': 10
        }
        
        try: 
            respuesta = requests.get(base_url, params=params, timeout=10)
            
            #Lanza error si el codigo es: 4XX 0 5XX
            respuesta.raise_for_status()
            
            datos = respuesta.json()
            
            if not datos:
                print(f" Pagina {pagina}: sin datos - fin de resultados")
                break
            
            todos_registros.extend(datos)
            print(f' Página {pagina}: {len(datos)} registros | acumlado: {len(todos_registros)}')
            
            if pagina < max_paginas:
                time.sleep(pausa_entre_pagina)
            
        except requests.exceptions.ConnectionError:
            print(f" Sin conexión a internet. Saltando extracción de la API.")
            break
        
        except requests.exceptions.Timeout:
            print(f" Timeout en página {pagina}. El Serivodr no respondió")
            break
        
        except requests.exceptions.HTTPError as e:
            print(f" Error HTTP {respuesta.status_code}: {e}")
            break
        
    if not todos_registros:
        return pd.DataFrame()
    
    return pd.json_normalize(todos_registros)

print("=== Extrayendo datos de JSONPlaceholder API ===")
df_api = extract_api(
    base_url='https://jsonplaceholder.typicode.com/users',
    max_paginas=2,
    pausa_entre_pagina=0.3
)

if not df_api.empty:
    print(f" API: {df_api.shape[0]} registros x {df_api.shape[1]} columnas")
    
    print("Columnas:", df_api.columns.tolist())
    display(df_api[['id', 'name', 'email', 'phone', 'company.name']].head(8))
    
else:
    print("Sin datos de API - continuamos con el CSV para las siguientes secciones.")
            
        
        
```
