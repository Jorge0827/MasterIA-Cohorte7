## ¿Qué es el Feature Engineering y por qué es crítico?


Es el proceso de transformar los datos crudos en variables que un modelo de ML puede entender y aprovechar.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder, OrdinalEncoder
from sklearn.feature_selection import SelectKBest, f_regression, f_classif
```

```python
df = pd.read_csv("SampleSuperstore.csv", encoding='latin1')

cols_utiles = ['Ship Mode', 'Segment', 'Region', 'Category', 'Sub-Category', 'Sales', 'Quantity', 'Discount', 'Profit']

df = df[cols_utiles].copy()

df.head(5)

```

## Encoding de variables categóricas..

Label Encoding = Un número por categoría (0,1,2) -----> 2 > 1 > 0

One Hot Encoding = Una columna binaria por categoría

Ordinal Encoding = Un número por categoría respetando el orden real ----> Categoría definida

```python
#Label Encoding, Asigna un número enteero por cata categoría.

#¿Cuando es correcto es usarlo? (Malo < Regular < Bueno) o prioridad (Baja < Media < Alta)

l_encoding = LabelEncoder()

#Resultado: Central->0    , East->1        , South->2          ,west->3
df['Region_label'] = l_encoding.fit_transform(df['Region'])


mapeo_label = dict(zip(l_encoding.classes_,l_encoding.transform(l_encoding.classes_)))

#Resultado o Mapeo

print('Mapeo Label Encoding - Region:', mapeo_label)
print()
print('Problema: El modelo va a interpretar West(3) > South(2) > East(1) > Central(0)')


```

```python
#Ordinal Encoding -> Este también produce un númeor por categoría, pero permite especificar el orden correcto tú mismo

orden_envio = [['Standard Class', 'Second Class', 'First Class', 'Same Day']]

or_encoder = OrdinalEncoder(categories=orden_envio)

df['Ship_Mode_ordinal'] = or_encoder.fit_transform(df[['Ship Mode']]).reshape(-1)

#Verfiy

df[['Ship Mode', 'Ship_Mode_ordinal']].drop_duplicates().sort_values('Ship_Mode_ordinal')


```

```python
#One Hot Encoding con pd.get_dummies = Es crear una columna binaria por cada categoría. Si un registro pertenece a esa categoria, la columna vale = 1, sino vale = 0.

df_ohe = pd.get_dummies(
    df,
    columns=['Segment', 'Region', 'Category'],
    drop_first=True, #Eliminar la primera categoría de cada variable
    dtype=int          #binario
)

cols_nuevas = [c for c in df_ohe.columns if c.startswith(('Segment_', 'Region_', 'Category_'))]

df_ohe[cols_nuevas].head(8)
```
